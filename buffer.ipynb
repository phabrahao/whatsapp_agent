```python

import requests
import base64
from PIL import Image
from io import BytesIO
from cryptography.hazmat.primitives.ciphers import Cipher, algorithms, modes
from cryptography.hazmat.primitives import hashes, hmac
import struct

def decrypt_whatsapp_media(encrypted_data, media_key_b64):
    """Decrypt WhatsApp media using the mediaKey"""
    
    # Decode media key
    media_key = base64.b64decode(media_key_b64)
    
    # WhatsApp uses specific key derivation
    # For images: use "WhatsApp Image Keys" + media_key
    info = b"WhatsApp Image Keys"
    
    # HKDF-like key derivation
    def hkdf_expand(key, info, length):
        h = hmac.HMAC(key, hashes.SHA256())
        h.update(info)
        h.update(b'\x01')
        return h.finalize()[:length]
    
    # Derive encryption key (32 bytes) and IV (16 bytes)
    expanded = hkdf_expand(media_key, info, 112)
    iv = expanded[:16]
    cipher_key = expanded[16:48]
    
    # Decrypt using AES-CBC
    cipher = Cipher(algorithms.AES(cipher_key), modes.CBC(iv))
    decryptor = cipher.decryptor()
    
    # Remove the last 10 bytes (MAC) before decryption
    ciphertext = encrypted_data[:-10]
    
    # Decrypt
    decrypted = decryptor.update(ciphertext) + decryptor.finalize()
    
    # Remove PKCS7 padding
    padding_length = decrypted[-1]
    return decrypted[:-padding_length]


    
# Download encrypted data
encrypted_data = requests.get(url).content

# Decrypt
image_data = decrypt_whatsapp_media(encrypted_data, media_key)

# Create PIL image
Image.open(BytesIO(image_data))
```

```python
def find_peaks(nums):
  if type(nums) != list:
    raise TypeError("input should be a list")
  for i in nums:
    if type(i) != int:
      raise TypeError("elements of the list should be a integer")
  if len(nums) == 0:
    return []
  solution = []
  if len(nums) == 1:
    return nums
  if nums[0] >= nums[1]:
    solution.append(nums[0])
  for i in range(1,len(nums)-1):
    if nums[i] >= nums[i+1] and nums[i] >= nums[i-1]:
      solution.append(nums[i])
  if nums[-1] >= nums[-2]:
    solution.append(nums[-1])
  return solution
```

```python
from Crypto.Cipher import AES
import hashlib
import hmac
import base64
import requests
from PIL import Image
from io import BytesIO

def decrypt_image(url, media_key):
    # Download encrypted data
    data = requests.get(url).content
    
    # Decode media key and expand with proper HKDF
    key = base64.b64decode(media_key)
    key = hmac.new(b"\0"*32, key, hashlib.sha256).digest()
    
    # HKDF expansion for 112 bytes
    keyStream = b""
    keyBlock = b""
    blockIndex = 1
    while len(keyStream) < 112:
        keyBlock = hmac.new(key, keyBlock + b"WhatsApp Image Keys" + chr(blockIndex).encode(), hashlib.sha256).digest()
        blockIndex += 1
        keyStream += keyBlock
    
    # Extract IV and cipher key
    iv = keyStream[:16]
    cipher_key = keyStream[16:48]
    
    # Decrypt (remove MAC)
    cipher = AES.new(cipher_key, AES.MODE_CBC, iv)
    decrypted = cipher.decrypt(data[:-10])
    
    # Remove padding
    decrypted = decrypted[:-decrypted[-1]]
    
    return decrypted
```
